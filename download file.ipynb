```python
print("=" * 60)
print("=" * 20, " S S Restro & Cafe ", "=" * 20)
print("=" * 60)

print("\n\n")

name = input("Enter the Customer Name = ")
contact_no = int(input("Enter the Mobile Number = "))

print("\n\n")

print("=" * 24, " ** Menu Card ** ", "=" * 24)

print("1. Pizza       Rate = 250 / Unit")
print("2. Burger      Rate = 150 / Unit")
print("3. Pasta       Rate = 180 / Plate")
print("4. Coffee      Rate = 50 / Cup")
print("5. Cold Drink  Rate = 50 / Ltr")
print("6. Thali       Rate = 80 / Plate")

print("=" * 60)

# Menu
item_menu = {
    1: ("Pizza", 250),
    2: ("Burger", 150),
    3: ("Pasta", 180),
    4: ("Coffee", 50),
    5: ("Cold Drink", 50),
    6: ("Thali", 80)
}

print("\n\n")
print("*" * 27, " ** BILL ** ", "*" * 27)

bill = []
total_item = 0

# Order
while True:

    choice = int(input("Enter the Item Number (0 to finish): "))

    if choice == 0:
        break

    if choice not in item_menu:
        print("Invalid item number!")
        continue

    quantity = int(input("Enter the quantity: "))

    item_name = item_menu[choice][0]
    price = item_menu[choice][1]

    total_price = quantity * price

    # Store item
    bill.append((item_name, quantity, price, total_price))

    total_item += quantity

    print("\nItem     :", item_name)
    print("Quantity :", quantity)
    print("Price    :", price)
    print("Total    :", total_price)

#  Bill
print("\n")
print("=" * 60)
print(" " * 20, "** FINAL BILL **")
print("=" * 60)

print("Customer Name :", name)
print("Contact No    :", contact_no)

print("=" * 60)

grand_total = 0

for item in bill:

    print(
        item[0],
        "| Qty:", item[1],
        "| Rate:", item[2],
        "| Total:", item[3]
    )

    grand_total += item[3]

print("-" * 60)

print("Total Items :", total_item)
print("Grand Total :", grand_total)

print("=" * 60,"\n")

print("="*25, "** Payment  OPtion **","="*25,"\n\n")

print("1. Cash")
print("2. UPI & QR")
print("3. Net Banking & Card ")
print("4. Cuppon")

payment=input("Enter your choice")
if payment == 1:
    print(" Cash " )
    
if payment == 2:
    print(" UPI  ")
    
if payment == 3:
    print(" Net Banking & Card ")
    
if payment == 4 :
    print(" Cuppon " )
    
    
import qrcode
from PIL import Image

upi_id = "shravan.kumar.yadav5@pingpay" 

upi_link = (
    f"upi://pay?"
    f"pa={upi_id}"
    f"&pn=Shravan%20Kumar"
    f"&am={grand_total}"
    f"&cu=INR"
)

qr = qrcode.make(upi_link)

print("Grand Total : ₹", grand_total)
print("Payment Method : UPI")
print("Please Scan QR Code")

display(qr)
    
print("          Thank You - Visit Again")

print("\n\n")
print("=" * 60)
```
